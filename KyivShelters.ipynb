```python
filepath = 'drive/MyDrive/kyiv_shelters.json'

with open(filepath, encoding="utf-8") as f:
    data = json.load(f)

data
```

```python
def nested(record, key):
    obj = record.get(key)
    if isinstance(obj, dict):
        return obj.get("imya")
    return None
```

```python
rows = []
for d in data:
    street = d.get("nazvaVulytsi") or ""
    number = d.get("inshiRekvizytyAdresy") or ""
    address = f"{street}, {number}".strip(", ")

    rows.append({
        "id":                   d.get("id"),

        # Location & Capacity
        "district":             nested(d, "NazvaRayonuMista"),
        "address":              address,
        "lat":                  d.get("shyrota"),
        "lon":                  d.get("dovhota"),
        "capacity":             d.get("mistkistOsib"),
        "area_m2":              d.get("ploshcha"),

        # Shelter Types
        "shelter_type":         nested(d, "TypSporudy"),
        "shelter_kind":         nested(d, "VydSporudy"),
        "location_type":        nested(d, "VidnosneRozmishchennya"),
        "functional_purpose":   nested(d, "FunktsionalnePryznachennya"),

        # Condition
        "readiness_status":     nested(d, "StanHotovnosti"),
        "always_ready":         d.get("perebuvayeUPostiyniyHotovnosti"),
        "technical_condition":  nested(d, "TekhnichnyyStan"),
        "sanitary_condition":   nested(d, "SanitarnyyStan"),
        "water":                d.get("systemaVodopostachannya"),
        "heating":              d.get("systemaOpalennya"),
        "power":                d.get("systemaEletrozhyvlennya"),
        "communication":        nested(d, "NayavniZasobyZvyazku"),
        "entrances":            d.get("kilkistVkhodivVykhodiv"),
        "emergency_exits":      d.get("kilkistAvariynykhVykhodiv"),

        # Accessibility
        "accessible_mgn":       d.get("nayavnistDostupuMalomobilnykhVerstvNaselennya"),

        # Availability
        "open_access":          nested(d, "BezpereshkodnyyDostup"),
        "ownership":            nested(d, "FormaVlasnosti"),
    })
```

```python
df = pd.DataFrame(rows)

print(f"Rows: {len(df)}, Columns: {len(df.columns)}")
print(df.dtypes)
print("\nSample")
print(df.head(3))
```

## Стан енергопостачання, водопостачання, опалення та зв'язок

```python
def clean_utility_status(text):
    if pd.isna(text) or text is None:
        return "Невідомо"

    text_clean = str(text).strip().lower()

    bad_keywords = ["відсутн", "не передбачен", "немає", "несправн", "відключ",
                    "доступ відсутній", "-", "не предбач", "не пердбач", "не має",
                    "не застосовується", "не визначалась"]

    if any(k in text_clean for k in bad_keywords):
        return "Відсутня/Несправна"

    good_keywords = ["наяв", "централ", "передбач", "мереж", "забезпеч", "справн",
                     "електр", "зовнішн", "працює", "водян", "радіатор", "резервуар",
                     "баки", "бутл", "ємност", "укомплектов", "пердбач", "бутильован", "ц"]

    if any(k in text_clean for k in good_keywords):
        return "Наявна/Справна"

    return "Інше/Невідомо"
```

```python
df['clean_water'] = df['water'].apply(clean_utility_status)
df['clean_heating'] = df['heating'].apply(clean_utility_status)
df['clean_power'] = df['power'].apply(clean_utility_status)

print("Водопостачання після очищення:")
print(df['clean_water'].value_counts())
```

### Забезпеченість районів водою

```python
water_counts = pd.crosstab(df['district'], df['clean_water'])

water_pcts = pd.crosstab(df['district'], df['clean_water'], normalize='index') * 100

water_report = pd.DataFrame(index=water_counts.index)
water_report['Всього укриттів (abs)'] = df['district'].value_counts()

water_report['Наявна/Справна (abs)'] = water_counts['Наявна/Справна']
water_report['Наявна/Справна (%)']   = water_pcts['Наявна/Справна'].round(1)

water_report['Відсутня/Несправна (abs)'] = water_counts['Відсутня/Несправна']
water_report['Відсутня/Несправна (%)']   = water_pcts['Відсутня/Несправна'].round(1)

water_report['Невідомо (abs)'] = water_counts['Невідомо']
water_report['Інше/Невідомо (abs)'] = water_counts['Інше/Невідомо']


water_report = water_report.loc[water_pcts.sort_values(by='Наявна/Справна', ascending=False).index]

water_report = water_report.reset_index().rename(columns={'district': 'Район міста'})

water_report
```

#### Зв'язок

```python
def clean_communication(text):
    if pd.isna(text) or text is None:
        return "Невідомо"

    text_clean = str(text).strip().lower()

    if "несправна" in text_clean:
        return "Зв'язок відсутній"

    if "справна" in text_clean and ("wi-fi" in text_clean or "провідна" in text_clean):
        return "Є інтернет (Wi-Fi/Дротовий)"

    if "відсутні" in text_clean:
        return "Зв'язок відсутній"

    return "Інше/Невідомо"

df['clean_communication'] = df['communication'].apply(clean_communication)

print("--- Розподіл засобів зв'язку в Києві ---")
df['clean_communication'].value_counts()
```

```python
comm_counts = pd.crosstab(df['district'], df['clean_communication'])

comm_pcts = pd.crosstab(df['district'], df['clean_communication'], normalize='index') * 100

comm_report = pd.DataFrame(index=comm_counts.index)

comm_report['Всього укриттів (abs)'] = df['district'].value_counts()

comm_report['Є інтернет (abs)'] = comm_counts['Є інтернет (Wi-Fi/Дротовий)']
comm_report['Є інтернет (%)']   = comm_pcts['Є інтернет (Wi-Fi/Дротовий)'].round(1)


comm_report['Зв\'язок відсутній (abs)'] = comm_counts['Зв\'язок відсутній']
comm_report['Зв\'язок відсутній (%)']   = comm_pcts['Зв\'язок відсутній'].round(1)


comm_report['Невідомо (abs)'] = comm_counts['Невідомо']

comm_report = comm_report.loc[comm_pcts.sort_values(by='Є інтернет (Wi-Fi/Дротовий)', ascending=False).index]

comm_report = comm_report.reset_index().rename(columns={'district': 'Район міста'})
print("=== СТАН ІНТЕРНЕТ-ЗВ'ЯЗКУ В УКРИТТЯХ ЗА РАЙОНАМИ ===")
comm_report
```
